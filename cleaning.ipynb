```python
import numpy as np
import pandas as pd

```

```python
df = pd.read_csv("crop_yield.csv")
```

```python
df.columns = df.columns.str.strip()
```

```python
# Strip whitespace from string columns
str_cols = df.select_dtypes(include='object').columns
df[str_cols] = df[str_cols].apply(lambda x: x.str.strip())

```

```python
# Drop duplicates 
df.drop_duplicates(inplace=True)
```

```python
df.fillna({
        'Crop': 'Unknown',
        'Season': 'Unknown',
        'State': 'Unknown',
        'Area': df['Area'].median(),
        'Production': df['Production'].median(),
        'Annual_Rainfall': df['Annual_Rainfall'].median(),
        'Fertilizer': df['Fertilizer'].median(),
        'Pesticide': df['Pesticide'].median(),
        'Yield': df['Yield'].median()
        }, inplace=True)
```

```python
# Remove extreme outliers in Yield
q_low = df['Yield'].quantile(0.01)
q_high = df['Yield'].quantile(0.99)
df = df[(df['Yield'] >= q_low) & (df['Yield'] <= q_high)]
```

```python
df.to_csv("cleaned_crop_data.csv", index=False)
```
